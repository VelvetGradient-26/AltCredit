```python
import pandas as pd
```

```python
# 1. Load the individual datasets
demo_df = pd.read_json('../data/demographic_data.json')
mapping_df = pd.read_json('../data/id_mapping.json')
merged_df = pd.read_json('../data/merged_data.json')

# 2. Identify unique columns to prevent duplicates (like user_id_x, user_id_y)
demo_cols_to_use = demo_df.columns.difference(merged_df.columns).tolist() + ['user_id']
mapping_cols_to_use = mapping_df.columns.difference(merged_df.columns).tolist() + ['user_id']

# 3. Merge everything into a single DataFrame
final_df = merged_df.copy()
final_df = final_df.merge(demo_df[demo_cols_to_use], on='user_id', how='left')
final_df = final_df.merge(mapping_df[mapping_cols_to_use], on='user_id', how='left')

# 4. Export to a single CSV file
final_df.to_csv('final_hackathon_dataset.csv', index=False)
```
